```python
operator= input("Enter an operator +,-,*,/ :")
if(operator not in ('+','-','*','/')):
    print("invalid operator")
else:    
    num1=int(input("Enter a number1 :"))
    num2=int(input("Enter a number2 "))
    
    if(operator=='+'):
       sum=num1 + num2
       print(sum)
    elif(operator=='-'):
        sum=num1-num2
        print(sum)
    elif(operator=='*'):
        sum=num1*num2
        print(sum)
    elif(operator=='/'):
         if(num2==0):
            print("num2 should be greater than 0")
         else:
            sum=num1/num2
            print(sum)
    
```
